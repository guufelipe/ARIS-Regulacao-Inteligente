## Tabela Solicitações

```python
# Definição das colunas do DataFrame de solicitações

colunas_solicitacoes = [
    'Decisao_Final',               # Rótulo (Y) - Status final: Regulado/Rejeitado
    'Data_Solicitacao',            # Data/Hora da submissão
    'HUF_Destino',                 # Hospital de Destino (HC-UFPE)
    'Especialidade',               # Especialidade (GASTROENTEROLOGIA)
    'Idade_Anos',                  # Idade do paciente (Numérica)
    'CID_Principal',               # Código CID-10 principal
    'Texto_Justificativa',         # Campo de texto livre para NLP
    'Sinais_Vitais_Brutos',        # Sinais vitais em texto/estrutura bruta
    'Exames_Laboratoriais',        # Resultados lab. em texto/estrutura bruta
    'Tempo_Regulacao_Minutos',     # Métrica de impacto
    'Estabelecimento_Solicitante', # Hospital Solicitante
    'Tipo_Internacao_Carater'      # Caráter (URGÊNCIA/ELETIVO)
]
```
